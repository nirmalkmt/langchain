Prompt types


```python
prompt = [
    { "role" : "system" , "content"  : "you are a python developer"},
    {"role" : "user" , "content" : "what is a list?"}
]
```

```python
from dotenv import load_dotenv  #load the environment variables
load_dotenv()
```

```python
from langchain_groq import ChatGroq
llm = ChatGroq(model = "openai/gpt-oss-120b")
```

```python
res = llm.invoke(prompt)
```

```python
print(res.content)
```

dynamic prompts

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
prompt = ChatPromptTemplate.from_messages([
    { "role" : "system" , "content"  : "you are a language translator , you will translate the text to the {language} "},
    {"role" : "user" , "content" : "{query}"}
])
```

```python
final_prompt = prompt.format_prompt(language = " english punjabi" , query = "are you a good boy?")
```

```python
res = llm.invoke(final_prompt.to_messages())
print(res.content)
```

Chaining


```python
chain = prompt | llm
```

```python

res = chain.invoke({
    "language": "English Punjabi",
    "query": "why are you doing like this?"
})
print(res.content)
```

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
out = StrOutputParser()
```

```python
out.invoke(res)
```

```python

chain = prompt | llm | out
```

```python
res = chain.invoke({
    "language": "Transliterate Punjabi",
    "query": "why are you doing like this?"
})
res
```

```python
def typeconversion(res):
    return res.upper()
```

```python
chain = prompt | llm | out | typeconversion
```

```python
res  = chain.invoke({
    "language": "Transliterated Punjabi",    
    "query": "when u will go to home?"
})
res
```
