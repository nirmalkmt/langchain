```python
#IMPLEMENTING THE CHAT GROQ MODEL
from dotenv import load_dotenv
load_dotenv()  # Load environment variables from .env file
```

 Import the ChatGroq class from the langchain_groq module  


```python
from langchain_groq import ChatGroq 
```

Create an instance of the ChatGroq class and invoke the llm


```python
llm  = ChatGroq(model= "openai/gpt-oss-120b",temperature = 0.9) 
prompt = input("Enter your prompt: ")
res = llm.invoke(prompt)
print(res.content)  
```
